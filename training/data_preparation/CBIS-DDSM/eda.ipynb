```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd

from skp.toolbox.functions import draw_bounding_boxes
```

```python
ROOT = "/mnt/stor/datasets/CBIS-DDSM/"
```

```python
df = pd.read_csv(os.path.join(ROOT, "cleaned_cbis_ddsm.csv"))
df
```

```python
file = np.random.choice(df.filepath.unique())

tmp_df = df.loc[df.filepath == file]
img = cv2.imread(os.path.join(ROOT, "png", file), cv2.IMREAD_GRAYSCALE)

plt.imshow(img, cmap="gray")
plt.show()

img_box = draw_bounding_boxes(img, list(tmp_df[["y1", "x1", "y2", "x2"]].values))
plt.imshow(img_box)
plt.show()
```
